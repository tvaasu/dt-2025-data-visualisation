```python
sentence = "minaolensinusober"
#extract first 3 letters
print("Print first 3 letters:", sentence[0:3])
print (sentence[:3])
# Extract letters from 5-9
print (sentence[5:9])
#extract letters from the 5th to end
print (sentence [5:])
#extract the last character 
print(sentence[-1])
#extract last 4 characters 
print (sentence[-4:])
#extract first letter of the substring starting from 5
#sentence[5:]: will extract a substring starting from 5
#[0]: this will take the 1st character of that substring
print(sentence[5:][0])
#extract first 5 letters of the substring starting from 5
print(sentence[5:][:5])
#extract entire string/text
print(sentence[:])
#extract every 2nd character
print(sentence[::2])
```

```python
sentence =  "I sometimes dont want to do standup but I still love it"

words = sentence.split(" ") #split the sentence to words

print(words)
#extract first word
print(words[0])
#extract the first 3 words
print(words[0:3])
#extract the last 3 words
print(words[-3:])
#print number of words
print("The number of words:", len(words))
```
