```python
import pandas as pd

df= pd.read_csv("movieslist.csv")
df.head()
```

```python
df.describe() #generate some statistics
```

```python
import matplotlib.pyplot as plt

df['Genre'].value_counts().plot(kind='bar')
plt.title('Movies by Genre')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```

The column chart shows that some genres like Drama and Horror appear much more frequently than others.

```python
df['Genre'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Movie Genres')
plt.ylabel('')
plt.show()
```

The pie chart gives a quick overview of how different genres share the overall dataset.
It confirms that a few genres take up most of the space — suggesting the dataset isn’t evenly distributed.
Visually, it’s easier to see which genres make up the largest portions compared to the bar chart.

```python
plt.scatter(df['Rating'], 
            df['Box Office (millions)'], 
            s=df['Duration (minutes)'], 
            alpha=0.6, c=df['Year'], cmap='viridis')

plt.title('Box Office vs Rating (Bubble size = Duration, Color = Year)')
plt.xlabel('Rating')
plt.ylabel('Box Office (millions)')
plt.colorbar(label='Year')
plt.show()
```

This scatter plot explores whether higher-rated movies tend to earn more money.
In this dataset, there isn’t a clear linear trend — some movies with average ratings still earned high box office numbers, while some highly rated ones didn’t make as much.
Bubble sizes show that longer movies (larger circles) don’t necessarily earn more, but there might be subtle patterns depending on the year.

```python
import pandas as pd

df = pd.read_csv('movieslist.csv')  # if file is in the same folder as your notebook

df.head()
```

```python
print(df.shape)
print(df.columns)
```

```python
import matplotlib.pyplot as plt

df.boxplot(column='Box Office (millions)', by='Genre', grid=False)
plt.title('Box Office Distribution by Genre')
plt.suptitle('')
plt.xlabel('Genre')
plt.ylabel('Box Office (millions)')
plt.xticks(rotation=45)
plt.show()
```

The box plot highlights how much box office earnings vary by genre.
Action and Adventure genres might have higher medians or wider ranges, suggesting bigger commercial success potential but also more variability.
