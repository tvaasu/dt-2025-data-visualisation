```python
text = """ I study at university and try to balance work, classes, and personal life.
Some days I feel motivated and focused, while other days I struggle with procrastination.
Technology and programming are a big part of my daily routine and learning process.
Python helps me automate tasks and understand data in a more practical way.
Overall, learning is a continuous process that involves mistakes, patience, and growth.
"""
```

```python
# convert text to lowercase and split into words
words = text.lower().split()

# show first 10 words
print("First 10 words:", words[:10])

# count total words
print("Total number of words:", len(words))
```

Task 1: Reading and processing text

In this task, I created a short text and stored it in a Python variable.
The text was converted to lowercase and split into individual words.
I printed the first ten words and calculated the total number of words
to demonstrate basic text processing in Python.


```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    stopwords={"and", "to", "at", "a", "the"}
).generate(text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

Task 2 Word cloud interpretation

The word cloud visualizes the most frequently used words in the text.
Words related to learning, university life, and Python appear larger,
indicating that these are the main themes of the text.
This visualization makes it easy to understand the focus of the text.
